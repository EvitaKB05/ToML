# Импорт и первичный анализ исходного датасета

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
# Вариант 12 -> задача №2 -> набор данных №4 (Heart Disease Dataset)
df = pd.read_csv('heart.csv')

print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

# Создание пропусков

```python
np.random.seed(42)
df_nan = df.copy()

# количественный признак: thalach (макс. ЧСС)
# категориальный признак: cp (тип боли в груди)
quant_col = 'thalach'
cat_col = 'cp'

n_missing = int(len(df_nan) * 0.1)
idx_quant = np.random.choice(df_nan.index, n_missing, replace=False)
idx_cat = np.random.choice(df_nan.index, n_missing, replace=False)

df_nan.loc[idx_quant, quant_col] = np.nan
df_nan.loc[idx_cat, cat_col] = np.nan

print(f"Пропусков в {quant_col}: {df_nan[quant_col].isna().sum()}")
print(f"Пропусков в {cat_col}: {df_nan[cat_col].isna().sum()}")
```

# Задача №2. Обработка пропусков

```python
df_clean = df_nan.copy()

# количественный признак - заполняем медианой
med_val = df_clean[quant_col].median()
df_clean[quant_col] = df_clean[quant_col].fillna(med_val)

# категориальный признак - заполняем модой
mode_val = df_clean[cat_col].mode()[0]
df_clean[cat_col] = df_clean[cat_col].fillna(mode_val)

print("Пропусков после обработки:", df_clean.isna().sum().sum())
```

# **Задача №2**
# Использованные методы:
- Для количественного признака (thalach) - заполнение медианой
- Для категориального признака (cp) - заполнение модой

# Признаки для построения моделей:
## Будем использовать все признаки, так как пропуски обработаны. 
Наиболее важными, судя по корреляции, будут cp, thalach, exang, oldpeak.

# Для более подробного и детального анализа и работы с датасетом выполним дополнительно задачу № 1
# Задача №1. Корреляционный анализ

```python
df_corr = df_nan.dropna()
corr_matrix = df_corr.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Матрица корреляции')
plt.tight_layout()
plt.show()
```

# Корреляция с целевой переменной

```python
target_corr = corr_matrix['target'].sort_values(ascending=False)
print(target_corr)
```

# **Задача №1**

# Корреляционный анализ показал:
- Наибольшая положительная корреляция с target: cp (0.43), thalach (0.42), slope (0.35)
- Наибольшая отрицательная корреляция: exang (-0.44), oldpeak (-0.43), ca (-0.39)

# Модели машинного обучения можно строить, так как есть признаки с умеренной корреляцией. 
## Мультиколлинеарность отсутствует (нет пар с корреляцией > 0.8).

# Задача №3. Масштабирование

```python
df_scaled = df.dropna().copy()

# масштабируем age
scaler = StandardScaler()
df_scaled['age_scaled'] = scaler.fit_transform(df_scaled[['age']])

print("Исходный age - среднее: {:.2f}, std: {:.2f}".format(df_scaled['age'].mean(), df_scaled['age'].std()))
print("Масштабированный - среднее: {:.2f}, std: {:.2f}".format(df_scaled['age_scaled'].mean(), df_scaled['age_scaled'].std()))
```

# Кодирование категориальных признаков

```python
# Label Encoding
le = LabelEncoder()
df_scaled['cp_label'] = le.fit_transform(df_scaled['cp'])

# One Hot Encoding
cp_dummies = pd.get_dummies(df_scaled['cp'], prefix='cp', dtype=int)
df_scaled = pd.concat([df_scaled, cp_dummies], axis=1)

print("Label Encoding:", df_scaled['cp_label'].unique())
print("One Hot Encoding:", cp_dummies.columns.tolist())
```

# **Задача №3**

# Масштабирование: использовали StandardScaler для признака age. Этот метод приводит данные к нулевому среднему и единичной дисперсии, что важно для алгоритмов, чувствительных к масштабу.

# Кодирование:
- Label Encoding - преобразует категории в числа 0,1,2... подходит для порядковых признаков
- One Hot Encoding - создает бинарные колонки для каждой категории, лучше для номинативных признаков, чтобы не создавать ложный порядок

# Задача №4. Гистограммы

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak', 'ca']

for i, col in enumerate(num_cols):
    axes[i].hist(df[col].dropna(), bins=20, edgecolor='black', alpha=0.7)
    axes[i].set_title(col)
    axes[i].set_xlabel(col)

plt.tight_layout()
plt.show()
```

# Boxplot

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    axes[i].boxplot(df[col].dropna())
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

# Диаграмма рассеяния

```python
plt.figure(figsize=(10, 6))
scatter = plt.scatter(df['age'], df['thalach'], c=df['target'], cmap='RdYlGn', alpha=0.6)
plt.xlabel('Возраст')
plt.ylabel('Макс. ЧСС')
plt.colorbar(scatter, label='Болезнь (1-есть)')
plt.grid(alpha=0.3)
plt.show()
```
