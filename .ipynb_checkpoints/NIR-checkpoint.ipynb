```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

plt.style.use('seaborn-v0_8-darkgrid')
```

```python
# Задание 1: загрузка данных
df = pd.read_csv('US_DOT_Airfare_Historical_2008_2025.csv')
print(f"Размер: {df.shape}")
df.head()
```

```python
# Задание 2: разведочный анализ
print("\nПропуски:")
print(df.isnull().sum().sum())

print("\nСтатистика целевой переменной:")
print(df['avg_fare'].describe())
```

```python
# графики
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0,0].hist(df['avg_fare'], bins=50, color='#4ECDC4', edgecolor='black')
axes[0,0].set_title('Распределение стоимости билетов')

axes[0,1].scatter(df['distance_miles'], df['avg_fare'], alpha=0.3, s=2, color='#4ECDC4')
axes[0,1].set_title('Стоимость vs Расстояние')
axes[0,1].set_xlabel('Мили')
axes[0,1].set_ylabel('Стоимость (USD)')

df.groupby('year')['avg_fare'].mean().plot(ax=axes[1,0], marker='o', color='#FF6B6B')
axes[1,0].set_title('Динамика по годам')

df.groupby('quarter')['avg_fare'].mean().plot(ax=axes[1,1], kind='bar', color='#4ECDC4')
axes[1,1].set_title('Сезонность')

plt.tight_layout()
plt.show()
```

```python
# Задание 3: feature engineering
df_fe = df.copy()

df_fe['quarter_sin'] = np.sin(2 * np.pi * df_fe['quarter'] / 4)
df_fe['quarter_cos'] = np.cos(2 * np.pi * df_fe['quarter'] / 4)
df_fe['year_quarter'] = df_fe['year'] + (df_fe['quarter'] - 1) / 4
df_fe['log_passengers'] = np.log1p(df_fe['passengers'])

print(f"Добавлено 4 признака. Всего: {len(df_fe.columns)}")
```

```python
# удаление признаков с утечкой данных
leaky = ['largest_carrier_fare', 'lowest_fare', 'lowest_fare_carrier_share']
df_fe = df_fe.drop(columns=[c for c in leaky if c in df_fe.columns])
print(f"Удалены: {leaky}")
print(f"Осталось признаков: {len(df_fe.columns)}")
```

```python
# Задание 4: корреляционный анализ
numeric_cols = df_fe.select_dtypes(include=[np.number]).columns.tolist()
corr = df_fe[numeric_cols].corr()['avg_fare'].sort_values(ascending=False)

print("\nКорреляция с avg_fare:")
print(corr)

plt.figure(figsize=(10, 6))
corr.drop('avg_fare').plot(kind='bar', color=['#4ECDC4' if x > 0 else '#FF6B6B' for x in corr.drop('avg_fare')])
plt.title('Корреляция признаков с целевой переменной')
plt.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# подготовка данных
X = df_fe.drop('avg_fare', axis=1)
y = df_fe['avg_fare']

categorical_features = X.select_dtypes(include=['object']).columns.tolist()
numeric_features = X.select_dtypes(include=[np.number]).columns.tolist()

preprocessor = ColumnTransformer([
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)
])

X_processed = preprocessor.fit_transform(X)
print(f"Признаков после one-hot: {X_processed.shape[1]}")

X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.2, random_state=42
)
print(f"Train: {X_train.shape[0]}, Test: {X_test.shape[0]}")
```

# Задание 5: выбор метрик (MAE, RMSE, R2)
# Метрики: MAE (средняя абсолютная ошибка), 
# RMSE (корень из среднеквадратичной), 
# R2 (коэффициент детерминации)

```python
# Задание 6: выбор моделей (5 моделей, 2 ансамблевых)
models = {
    'Linear Regression': LinearRegression(),
    'Decision Tree': DecisionTreeRegressor(random_state=42, max_depth=10),
    'KNN': KNeighborsRegressor(n_neighbors=10),
    'Random Forest': RandomForestRegressor(random_state=42, n_estimators=100, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(random_state=42, n_estimators=100)
}
```

```python
# Задание 7-8: baseline модели
print("\nBaseline models:")
print("="*50)

baseline_results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    results = {
        'Model': name,
        'MAE': mean_absolute_error(y_test, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
        'R2': r2_score(y_test, y_pred)
    }
    baseline_results.append(results)
    print(f"\n{name}:")
    print(f"  MAE: ${results['MAE']:.2f}")
    print(f"  R²: {results['R2']:.4f}")

baseline_df = pd.DataFrame(baseline_results).set_index('Model')
```

```python
# график baseline
fig, ax = plt.subplots(figsize=(10, 5))
baseline_df['MAE'].plot(kind='bar', ax=ax, color='#FF6B6B')
ax.set_title('Baseline: сравнение MAE')
ax.set_ylabel('Ошибка (USD)')
ax.set_xlabel('Модель')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Задание 9: подбор гиперпараметров
print("\nПодбор гиперпараметров:")
print("="*50)

param_grids = {
    'Linear Regression': {'fit_intercept': [True, False]},
    'Decision Tree': {'max_depth': [5, 10, 15], 'min_samples_split': [5, 10]},
    'KNN': {'n_neighbors': [5, 10, 15], 'weights': ['uniform', 'distance']},
    'Random Forest': {'n_estimators': [50, 100], 'max_depth': [10, 15], 'min_samples_split': [5, 10]},
    'Gradient Boosting': {'n_estimators': [50, 100], 'learning_rate': [0.05, 0.1], 'max_depth': [3, 5]}
}

best_models = {}
tuned_results = []

for name in models.keys():
    print(f"\n{name}...")
    
    grid = GridSearchCV(
        estimator=models[name],
        param_grid=param_grids[name],
        cv=3,
        scoring='neg_mean_absolute_error',
        n_jobs=-1,
        verbose=0
    )
    
    grid.fit(X_train, y_train)
    best_models[name] = grid.best_estimator_
    
    print(f"Лучшие параметры: {grid.best_params_}")
    print(f"Лучшее MAE на CV: ${-grid.best_score_:.2f}")
    
    y_pred = grid.predict(X_test)
    print(f"Тест MAE: ${mean_absolute_error(y_test, y_pred):.2f}")
    print(f"Тест R²: {r2_score(y_test, y_pred):.4f}")
    
    tuned_results.append({
        'Model': name,
        'MAE': mean_absolute_error(y_test, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
        'R2': r2_score(y_test, y_pred)
    })

tuned_df = pd.DataFrame(tuned_results).set_index('Model')
```

```python
# Задание 10: сравнение baseline и tuned
print("\nСравнение baseline и tuned")
print("="*50)

comparison = pd.DataFrame({
    'Baseline_MAE': baseline_df['MAE'],
    'Tuned_MAE': tuned_df['MAE'],
    'Improvement': baseline_df['MAE'] - tuned_df['MAE']
})
print(comparison.round(2))

fig, ax = plt.subplots(figsize=(10, 5))
x = np.arange(len(models))
width = 0.35

ax.bar(x - width/2, baseline_df['MAE'], width, label='Baseline', color='#FF6B6B')
ax.bar(x + width/2, tuned_df['MAE'], width, label='Tuned', color='#4ECDC4')
ax.set_xticks(x)
ax.set_xticklabels(models.keys(), rotation=45)
ax.set_ylabel('MAE (USD)')
ax.set_title('Сравнение ошибок моделей')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Задание 11: анализ лучшей модели
best_name = tuned_df['R2'].idxmax()
best_model = best_models[best_name]

print(f"\nЛучшая модель: {best_name}")
print(f"MAE: ${tuned_df.loc[best_name, 'MAE']:.2f}")
print(f"R²: {tuned_df.loc[best_name, 'R2']:.4f}")

y_pred = best_model.predict(X_test)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(y_test, y_pred, alpha=0.3, s=2, color='#4ECDC4')
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
axes[0].set_xlabel('Факт')
axes[0].set_ylabel('Предсказание')
axes[0].set_title(f'{best_name}: предсказания vs факт')

residuals = y_test - y_pred
axes[1].hist(residuals, bins=50, color='#4ECDC4', edgecolor='black')
axes[1].axvline(x=0, color='red', linestyle='--')
axes[1].set_xlabel('Ошибка (USD)')
axes[1].set_title('Распределение ошибок')

plt.tight_layout()
plt.show()
```

```python
# Задание 12: сохранение модели
import joblib

joblib.dump(best_model, 'best_model.pkl')
joblib.dump(preprocessor, 'preprocessor.pkl')
print("\nСохранено: best_model.pkl, preprocessor.pkl")
```
