# Цель работы
Изучение способов подготовки выборки и подбора гиперпараметров на примере метода ближайших соседей (KNN).

# Задачи
1. Подготовить данные Titanic для задачи классификации
2. Разделить данные на обучающую и тестовую выборки
3. Обучить KNN с произвольным параметром K
4. Выполнить подбор гиперпараметра K с помощью GridSearchCV и RandomizedSearchCV
5. Использовать не менее двух стратегий кросс-валидации
6. Сравнить качество исходной и оптимальной моделей

```python
# 2 импорт библиотек и датасета
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from scipy.stats import randint

# Загрузка данных Titanic из seaborn
df = sns.load_dataset('titanic')
print(f"Размер исходного датасета: {df.shape}")
```

```python
# 3 первичный анализ 
# копия
data = df.copy()

# Пропуски
# Возраст заполняем медианой по классу
age_medians = data.groupby('pclass')['age'].median()
def fill_age(row):
    if pd.isnull(row['age']):
        return age_medians[row['pclass']]
    return row['age']
data['age'] = data.apply(fill_age, axis=1)

# Порт посадки заполняем модой
data['embarked'].fillna(data['embarked'].mode()[0], inplace=True)

# Удаляем бесполезные столбцы
cols_to_drop = ['deck', 'embark_town', 'alive', 'class', 'who']
data.drop(columns=[col for col in cols_to_drop if col in data.columns], inplace=True)

# Кодирование категориальных признаков
data = pd.get_dummies(data, columns=['sex', 'embarked'], drop_first=True)

# Преобразование булевых в int
bool_cols = data.select_dtypes(include=['bool']).columns
for col in bool_cols:
    data[col] = data[col].astype(int)

print(f"Размер после предобработки: {data.shape}")
print(data.head())
```

```python
# 4 Целевая переменая
# Целевая переменная - survived
X = data.drop('survived', axis=1)
y = data['survived']

print(f"Признаки X: {X.shape}")
print(f"Целевая переменная y: {y.shape}")
print(f"Распределение классов:\n{y.value_counts()}")
```

```python
# 5 Масштабирование Кнн
# Масштабирование числовых признаков
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Преобразуем обратно в DataFrame для наглядности
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
print("Статистика после масштабирования:")
print(X_scaled.describe().round(2))
```

```python
# 6 Обучение и Тест
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Обучающая выборка: X_train {X_train.shape}, y_train {y_train.shape}")
print(f"Тестовая выборка: X_test {X_test.shape}, y_test {y_test.shape}")
print(f"Соотношение классов в обучающей:\n{y_train.value_counts(normalize=True).round(2)}")
```

```python
# 7 Модель с К = 5
# Обучаем модель с K=5
knn_default = KNeighborsClassifier(n_neighbors=5)
knn_default.fit(X_train, y_train)

# Предсказание на тесте
y_pred_default = knn_default.predict(X_test)

# Оценка качества
accuracy_default = accuracy_score(y_test, y_pred_default)
print(f"Accuracy модели с K=5: {accuracy_default:.4f}")

print("\nClassification report:")
print(classification_report(y_test, y_pred_default))

print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred_default))
```

```python
# 8 Кфолд и грид
# Создаем сетку параметров для перебора
param_grid = {'n_neighbors': range(1, 31, 2)}  # Нечетные от 1 до 30

# Стратегия 1: KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# GridSearchCV
grid_search = GridSearchCV(
    KNeighborsClassifier(), 
    param_grid, 
    cv=kf, 
    scoring='accuracy',
    verbose=1
)

print("Выполняется GridSearchCV с KFold...")
grid_search.fit(X_train, y_train)

print(f"\nЛучший параметр K (GridSearch + KFold): {grid_search.best_params_}")
print(f"Лучшая accuracy на кросс-валидации: {grid_search.best_score_:.4f}")
```

```python
# 9 RandomizedSearchCV с StratifiedKFold
# Стратегия 2: StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Распределение для случайного поиска
param_dist = {'n_neighbors': randint(1, 30)}

# RandomizedSearchCV
random_search = RandomizedSearchCV(
    KNeighborsClassifier(),
    param_dist,
    n_iter=15,
    cv=skf,
    scoring='accuracy',
    random_state=42,
    verbose=1
)

print("Выполняется RandomizedSearchCV с StratifiedKFold...")
random_search.fit(X_train, y_train)

print(f"\nЛучший параметр K (RandomizedSearch + StratifiedKFold): {random_search.best_params_}")
print(f"Лучшая accuracy на кросс-валидации: {random_search.best_score_:.4f}")
```

```python
# 10 Визуализация зависимости accuracy от K
# Получаем результаты GridSearchCV
results = pd.DataFrame(grid_search.cv_results_)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(results['param_n_neighbors'], results['mean_test_score'], marker='o')
plt.fill_between(results['param_n_neighbors'], 
                  results['mean_test_score'] - results['std_test_score'],
                  results['mean_test_score'] + results['std_test_score'], alpha=0.2)
plt.xlabel('K (количество соседей)')
plt.ylabel('Accuracy')
plt.title('Зависимость accuracy от K')
plt.grid(True)

# Сравнение методов поиска
plt.subplot(1, 2, 2)
methods = ['Default (K=5)', 'GridSearch', 'RandomizedSearch']
accuracy_values = [
    accuracy_default,
    grid_search.best_score_,
    random_search.best_score_
]
colors = ['red', 'green', 'blue']
plt.bar(methods, accuracy_values, color=colors, alpha=0.7)
plt.ylabel('Accuracy')
plt.title('Сравнение accuracy на кросс-валидации')
for i, v in enumerate(accuracy_values):
    plt.text(i, v + 0.005, f'{v:.4f}', ha='center')

plt.tight_layout()
plt.show()
```

```python
# Оптимальная модель и тестирование
# 11 Используем лучшую модель из GridSearchCV
knn_best = grid_search.best_estimator_

# Предсказание на тесте
y_pred_best = knn_best.predict(X_test)

# Оценка на тестовой выборке
accuracy_best = accuracy_score(y_test, y_pred_best)
print(f"Accuracy оптимальной модели (K={grid_search.best_params_['n_neighbors']}) на тесте: {accuracy_best:.4f}")

print("\nClassification report оптимальной модели:")
print(classification_report(y_test, y_pred_best))

print("Confusion matrix оптимальной модели:")
print(confusion_matrix(y_test, y_pred_best))
```

```python
# 12 Сравнение моделей
# Таблица сравнения
comparison = pd.DataFrame({
    'Модель': ['Произвольная (K=5)', f'Оптимальная (K={grid_search.best_params_["n_neighbors"]})'],
    'Accuracy на кросс-валидации': [np.nan, grid_search.best_score_],
    'Accuracy на тесте': [accuracy_default, accuracy_best]
})
print("Сравнение моделей:")
print(comparison)

# Проверка на переобучение
print("\nПроверка на переобучение:")
print(f"Произвольная модель - Кросс-валидация: N/A, Тест: {accuracy_default:.4f}")
print(f"Оптимальная модель - Кросс-валидация: {grid_search.best_score_:.4f}, Тест: {accuracy_best:.4f}")
if abs(grid_search.best_score_ - accuracy_best) < 0.05:
    print("Переобучения нет - результаты на кросс-валидации и тесте близки")
else:
    print("Возможно переобучение - есть расхождение между кросс-валидацией и тестом")
```

```python
# 13 Дополнительное сравнение с cross_val_score
# Оценим модель с помощью cross_val_score напрямую
cv_scores_default = cross_val_score(knn_default, X_train, y_train, cv=KFold(5, shuffle=True, random_state=42))
cv_scores_best = cross_val_score(knn_best, X_train, y_train, cv=KFold(5, shuffle=True, random_state=42))

print("Сравнение кросс-валидации на обучающей выборке:")
print(f"Произвольная модель (K=5): среднее={cv_scores_default.mean():.4f} (+/- {cv_scores_default.std()*2:.4f})")
print(f"Оптимальная модель (K={grid_search.best_params_['n_neighbors']}): среднее={cv_scores_best.mean():.4f} (+/- {cv_scores_best.std()*2:.4f})")

# Визуализация с правильным параметром
plt.figure(figsize=(10, 5))
box_plot = plt.boxplot([cv_scores_default, cv_scores_best], 
                       tick_labels=['K=5', f'K={grid_search.best_params_["n_neighbors"]}'])  # Исправлено: labels -> tick_labels
plt.ylabel('Accuracy')
plt.title('Сравнение результатов кросс-валидации')
plt.grid(True, alpha=0.3)
plt.show()

# Дополнительная статистика
print(f"\nУлучшение: +{(cv_scores_best.mean() - cv_scores_default.mean())*100:.2f}%")
```
