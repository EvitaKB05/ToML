## Постановка задачи

Найти экстремум функции:

\[
f(x) = 2(x_1^4 - x_2^2)^2 + 3x_1x_2 \rightarrow extr
\]

при ограничениях:

\[
\begin{cases}
g_1(x) = x_1^2 + x_2^2 - 6 \leq 0, \\
g_2(x) = -x_1 \leq 0.
\end{cases}
\]

Множество допустимых решений:

\[
X = \{ x \in \mathbb{R}^2 \mid g_1(x) \leq 0, \; g_2(x) \leq 0 \}
\]

## Теоретические сведения

### Обобщенная функция Лагранжа

\[
L(x, \lambda_0, \lambda) = \lambda_0 f(x) + \sum_{j=1}^{p} \lambda_j g_j(x)
\]

### Необходимые условия экстремума первого порядка

Пусть \(x^e\) – точка локального экстремума. Тогда существуют числа \(\lambda_0^e, \lambda_1^e, ..., \lambda_p^e\), не равные одновременно нулю, такие что:

1. **Условие стационарности:**
   \[
   \frac{\partial L(x^e, \lambda_0^e, \lambda^e)}{\partial x_i} = 0, \quad i = 1, \dots, n
   \]

2. **Условие допустимости:**
   \[
   g_j(x^e) = 0, \; j = 1,\dots,m; \quad g_j(x^e) \leq 0, \; j = m+1,\dots,p
   \]

3. **Условие неотрицательности (для минимума):**
   \[
   \lambda_j^e \geq 0, \quad j = m+1,\dots,p
   \]

4. **Условие дополняющей нежесткости:**
   \[
   \lambda_j^e g_j(x^e) = 0, \quad j = m+1,\dots,p
   \]

### Достаточные условия второго порядка

Если в стационарной точке \(x^e\) второй дифференциал классической функции Лагранжа положителен для всех ненулевых \(dx\), удовлетворяющих:
\[
dg_j(x^e) = 0 \quad (j \in J_{eq}), \quad dg_j(x^e) \leq 0 \quad (j \in J_a, \lambda_j^e = 0)
\]
то \(x^e\) – точка локального минимума (при \(d^2L > 0\)) или максимума (при \(d^2L < 0\)).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve, minimize
import warnings
warnings.filterwarnings('ignore')

def f(x1, x2):
    return 2 * (x1**4 - x2**2)**2 + 3 * x1 * x2

def g1(x1, x2):
    return x1**2 + x2**2 - 6

def g2(x1, x2):
    return -x1

stations = []

# 1: g1=0, x1>0, l2=0
def case1(v):
    x1, x2, l1 = v
    dL1 = 4*(x1**4 - x2**2)*(4*x1**3) + 3*x2 + 2*l1*x1
    dL2 = 4*(x1**4 - x2**2)*(-2*x2) + 3*x1 + 2*l1*x2
    g1v = x1**2 + x2**2 - 6
    return [dL1, dL2, g1v]

for x1g in [0.5, 1.0, 1.5, 2.0, 2.2, 2.4]:
    for x2g in [-2.5, -2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2, 2.5]:
        try:
            sol = fsolve(case1, [x1g, x2g, 0.0], maxfev=500)
            x1, x2, l1 = sol
            if x1 > 1e-6 and abs(x1**2 + x2**2 - 6) < 0.01:
                unique = True
                for s in stations:
                    if abs(s[0]-x1) < 0.01 and abs(s[1]-x2) < 0.01:
                        unique = False
                        break
                if unique:
                    stations.append((x1, x2, l1, 0, f(x1, x2)))
        except:
            pass

# 2: x1=0, g1<0, l2>0
def case2(v):
    x2, l1, l2 = v
    eq1 = 3*x2 - l2
    eq2 = 8*x2**3 + 2*l1*x2
    return [eq1, eq2]

for x2g in [-2.5, -2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2, 2.5]:
    try:
        sol = fsolve(case2, [x2g, 0.0, 1.0], maxfev=500)
        x2, l1, l2 = sol
        if x2**2 - 6 < -0.01 and l2 > 1e-6:
            unique = True
            for s in stations:
                if abs(s[0]-0) < 0.01 and abs(s[1]-x2) < 0.01:
                    unique = False
                    break
            if unique:
                stations.append((0, x2, l1, l2, f(0, x2)))
    except:
        pass

# 3: x1=0, g1=0
x2_vals = [np.sqrt(6), -np.sqrt(6)]
for x2 in x2_vals:
    l2 = 3 * x2
    l1 = -4 * x2**2 if abs(x2) > 1e-6 else 0
    if l2 >= -1e-6:
        unique = True
        for s in stations:
            if abs(s[0]-0) < 0.01 and abs(s[1]-x2) < 0.01:
                unique = False
                break
        if unique:
            stations.append((0, x2, l1, l2, f(0, x2)))

# 4: внутренняя точка (g1<0, x1>0, l1=0, l2=0)
def case4(v):
    x1, x2 = v
    dL1 = 4*(x1**4 - x2**2)*(4*x1**3) + 3*x2
    dL2 = 4*(x1**4 - x2**2)*(-2*x2) + 3*x1
    return [dL1, dL2]

for x1g in [0.1, 0.5, 1.0, 1.5]:
    for x2g in [-1.5, -1, -0.5, 0, 0.5, 1, 1.5]:
        try:
            sol = fsolve(case4, [x1g, x2g], maxfev=500)
            x1, x2 = sol
            if x1 > 1e-6 and x1**2 + x2**2 - 6 < -0.01:
                unique = True
                for s in stations:
                    if abs(s[0]-x1) < 0.01 and abs(s[1]-x2) < 0.01:
                        unique = False
                        break
                if unique:
                    stations.append((x1, x2, 0, 0, f(x1, x2)))
        except:
            pass

print("Стационарные точки:")
for i, (x1, x2, l1, l2, fv) in enumerate(stations):
    print(f"{i+1}: ({x1:.4f}, {x2:.4f}) -> f = {fv:.6f}")

if len(stations) > 0:
    min_idx = np.argmin([s[4] for s in stations])
    max_idx = np.argmax([s[4] for s in stations])
    print(f"\nМинимум: ({stations[min_idx][0]:.4f}, {stations[min_idx][1]:.4f}) -> f = {stations[min_idx][4]:.6f}")
    print(f"Максимум: ({stations[max_idx][0]:.4f}, {stations[max_idx][1]:.4f}) -> f = {stations[max_idx][4]:.6f}")

# численная проверка
def obj_num(x):
    return 2 * (x[0]**4 - x[1]**2)**2 + 3 * x[0] * x[1]

def constr_num(x):
    return 6 - (x[0]**2 + x[1]**2)

bounds = [(0, None), (None, None)]
constr = {'type': 'ineq', 'fun': constr_num}

verification = []
for x0 in [[0.5, 0.5], [1.0, 0], [0, 1.0], [2.0, 0], [0, 2.0], [1.5, 1.5], [2.0, 1.0], [0.5, -1.0]]:
    res = minimize(obj_num, x0, bounds=bounds, constraints=constr, method='slsqp')
    if res.success:
        found = False
        for v in verification:
            if abs(v[0]-res.x[0]) < 0.001 and abs(v[1]-res.x[1]) < 0.001:
                found = True
                break
        if not found:
            verification.append((res.x[0], res.x[1], res.fun))

print("\nЧисленная проверка:")
for x1, x2, fv in verification:
    print(f"  ({x1:.4f}, {x2:.4f}) -> f = {fv:.6f}")

# визуализация
x1_vals = np.linspace(-0.5, 3, 300)
x2_vals = np.linspace(-3, 3, 300)
X1, X2 = np.meshgrid(x1_vals, x2_vals)
F = 2 * (X1**4 - X2**2)**2 + 3 * X1 * X2
circle = X1**2 + X2**2 <= 6
F_masked = np.ma.masked_where(~circle, F)
F_masked = np.ma.masked_where(X1 < 0, F_masked)

plt.figure(figsize=(12, 10))
contour = plt.contour(X1, X2, F_masked, levels=30, cmap='viridis')
plt.colorbar(contour, label='f(x)')
circle_plot = plt.Circle((0, 0), np.sqrt(6), fill=False, edgecolor='red', linewidth=2)
plt.gca().add_patch(circle_plot)
plt.axvline(x=0, color='blue', linestyle='--', linewidth=2)

if len(stations) > 0:
    xs = [s[0] for s in stations]
    ys = [s[1] for s in stations]
    plt.scatter(xs, ys, color='red', s=100, zorder=5)
    for x, y, fv in zip(xs, ys, [s[4] for s in stations]):
        plt.annotate(f'{fv:.1f}', (x, y), xytext=(5, 5), textcoords='offset points', fontsize=9)

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('f(x) = 2(x1^4 - x2^2)^2 + 3 x1 x2')
plt.xlim(-0.5, 3)
plt.ylim(-3, 3)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.tight_layout()
plt.show()
```
