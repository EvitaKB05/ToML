```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.seasonal import seasonal_decompose
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split
from gplearn.genetic import SymbolicRegressor
import warnings
warnings.filterwarnings('ignore')

df = sns.load_dataset('flights')
df['month_num'] = pd.date_range(start='1949-01-01', periods=len(df), freq='MS').month
df['year'] = pd.date_range(start='1949-01-01', periods=len(df), freq='MS').year

print("Датасет flights загружен")
print(f"Размер: {df.shape}")
print(df.head())

plt.figure(figsize=(14, 6))
plt.plot(df['passengers'], color='blue', linewidth=1)
plt.xlabel('месяцы (1949-1960)')
plt.ylabel('количество пассажиров (тыс)')
plt.title('Временной ряд: авиапассажиры')
plt.grid(True, alpha=0.3)
plt.show()

decomposition = seasonal_decompose(df['passengers'], model='multiplicative', period=12)
fig, axes = plt.subplots(4, 1, figsize=(14, 10))
axes[0].plot(decomposition.observed, label='наблюдаемый ряд')
axes[0].legend()
axes[1].plot(decomposition.trend, label='тренд', color='green')
axes[1].legend()
axes[2].plot(decomposition.seasonal, label='сезонность', color='orange')
axes[2].legend()
axes[3].plot(decomposition.resid, label='остатки', color='red')
axes[3].legend()
plt.tight_layout()
plt.show()

train_size = int(len(df) * 0.8)
train = df.iloc[:train_size]
test = df.iloc[train_size:]

print(f"Обучающая выборка: {train_size} месяцев ({train['year'].min()}-{train['year'].max()})")
print(f"Тестовая выборка: {len(test)} месяцев ({test['year'].min()}-{test['year'].max()})")

model_arima = ARIMA(train['passengers'], order=(2, 1, 2), seasonal_order=(1, 1, 1, 12))
model_fit = model_arima.fit()
forecast_arima = model_fit.forecast(steps=len(test))

print("\nARIMA параметры:")
print(model_fit.summary().tables[1])

train_lags = train['passengers'].values
test_lags = test['passengers'].values

def create_lags(data, n_lags=12):
    X, y = [], []
    for i in range(n_lags, len(data)):
        X.append(data[i-n_lags:i])
        y.append(data[i])
    return np.array(X), np.array(y)

X_train, y_train = create_lags(train_lags, n_lags=12)
X_test, y_test = create_lags(test_lags, n_lags=12)

if len(X_train) > 0:
    sym_reg = SymbolicRegressor(population_size=1000, generations=20, 
                                 stopping_criteria=0.01, p_crossover=0.7,
                                 p_subtree_mutation=0.1, p_hoist_mutation=0.05,
                                 p_point_mutation=0.1, max_samples=0.9,
                                 verbose=0, random_state=42, function_set=('add', 'sub', 'mul', 'div'))
    sym_reg.fit(X_train, y_train)
    forecast_symbolic = sym_reg.predict(X_test)
else:
    print("Недостаточно данных для символьной регрессии")
    forecast_symbolic = np.full(len(test), np.mean(train['passengers']))

plt.figure(figsize=(14, 7))
plt.plot(train.index, train['passengers'], label='Обучающая выборка', color='blue')
plt.plot(test.index, test['passengers'], label='Реальные значения', color='black', linewidth=2)
plt.plot(test.index, forecast_arima, label='ARIMA прогноз', color='red', linestyle='--', linewidth=1.5)
if len(forecast_symbolic) > 0:
    plt.plot(test.index[:len(forecast_symbolic)], forecast_symbolic, label='Символьная регрессия', color='green', linestyle='--', linewidth=1.5)
plt.xlabel('месяцы')
plt.ylabel('количество пассажиров (тыс)')
plt.title('Прогнозы временного ряда')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

mse_arima = mean_squared_error(test['passengers'][:len(forecast_arima)], forecast_arima)
mae_arima = mean_absolute_error(test['passengers'][:len(forecast_arima)], forecast_arima)

print("\nОценка качества прогнозов")
print("="*50)
print(f"ARIMA:")
print(f"  MSE = {mse_arima:.4f}")
print(f"  MAE = {mae_arima:.4f}")

if len(forecast_symbolic) > 0:
    mse_sym = mean_squared_error(test['passengers'][:len(forecast_symbolic)], forecast_symbolic)
    mae_sym = mean_absolute_error(test['passengers'][:len(forecast_symbolic)], forecast_symbolic)
    print(f"\nСимвольная регрессия:")
    print(f"  MSE = {mse_sym:.4f}")
    print(f"  MAE = {mae_sym:.4f}")
    
    print("\nСравнение:")
    if mse_arima < mse_sym:
        print(f"Лучшая модель: ARIMA (MSE = {mse_arima:.4f})")
    else:
        print(f"Лучшая модель: символьная регрессия (MSE = {mse_sym:.4f})")

plt.figure(figsize=(14, 6))
plt.plot(test.index, test['passengers'], label='Реальные значения', color='black', linewidth=2)
plt.plot(test.index, forecast_arima, label='ARIMA', color='red', linestyle='--', linewidth=1.5)
if len(forecast_symbolic) > 0:
    plt.plot(test.index[:len(forecast_symbolic)], forecast_symbolic, label='Символьная регрессия', color='green', linestyle='--', linewidth=1.5)
plt.xlabel('месяцы')
plt.ylabel('количество пассажиров (тыс)')
plt.title('Сравнение прогнозов на тестовой выборке')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

if len(forecast_symbolic) > 0:
    print("\nПримечание: Модели МГУА не использовались из-за невозможности установки библиотеки gmdh")
    
```
