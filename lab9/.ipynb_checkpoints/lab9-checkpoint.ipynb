# Кудабаева Эвита ИУ5-62Б

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
import warnings
warnings.filterwarnings('ignore')

data = load_iris()
X_d1 = data.data
y_true = data.target

print(f"Датасет Iris загружен")
print(f"Размер D1: {X_d1.shape} (150 строк, 4 признака)")
print(f"Признаки: sepal length, sepal width, petal length, petal width")
print(f"Реальные классы (для проверки): {np.unique(y_true)}")

pca = PCA(n_components=2)
X_d2 = pca.fit_transform(X_d1)

print(f"\nPCA: объяснённая дисперсия = {pca.explained_variance_ratio_[0]:.4f} + {pca.explained_variance_ratio_[1]:.4f} = {sum(pca.explained_variance_ratio_):.4f}")

tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_d3 = tsne.fit_transform(X_d1)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].scatter(X_d2[:, 0], X_d2[:, 1], c=y_true, cmap='viridis', alpha=0.7)
axes[0].set_title('PCA: снижение размерности до 2')
axes[0].set_xlabel('PC1')
axes[0].set_ylabel('PC2')
axes[1].scatter(X_d3[:, 0], X_d3[:, 1], c=y_true, cmap='viridis', alpha=0.7)
axes[1].set_title('t-SNE: снижение размерности до 2')
axes[1].set_xlabel('t-SNE1')
axes[1].set_ylabel('t-SNE2')
plt.tight_layout()
plt.show()

print("\nНаиболее явно кластеры выделены на графике t-SNE")

clustering_methods = [
    ('K-Means', KMeans(n_clusters=3, random_state=42)),
    ('Agglomerative', AgglomerativeClustering(n_clusters=3)),
    ('DBSCAN', DBSCAN(eps=0.5, min_samples=5))
]

datasets = {
    'D1 (исходный, 4 признака)': X_d1,
    'D2 (PCA, 2 признака)': X_d2,
    'D3 (t-SNE, 2 признака)': X_d3
}

results = []

for dataset_name, X_data in datasets.items():
    print(f"\n{'='*60}")
    print(f"{dataset_name}")
    print('='*60)
    
    for method_name, model in clustering_methods:
        labels = model.fit_predict(X_data)
        n_clusters_found = len(np.unique(labels))
        
        if n_clusters_found > 1:
            sil = silhouette_score(X_data, labels)
            db = davies_bouldin_score(X_data, labels)
            ch = calinski_harabasz_score(X_data, labels)
        else:
            sil = -1
            db = float('inf')
            ch = 0
        
        results.append({
            'dataset': dataset_name,
            'method': method_name,
            'n_clusters': n_clusters_found,
            'silhouette': sil,
            'davies_bouldin': db,
            'calinski_harabasz': ch
        })
        
        print(f"\n{method_name}:")
        print(f"  найдено кластеров: {n_clusters_found}")
        if n_clusters_found > 1:
            print(f"  silhouette: {sil:.4f}")
            print(f"  davies_bouldin: {db:.4f}")
            print(f"  calinski_harabasz: {ch:.2f}")
        else:
            print(f"  недостаточно кластеров для оценки")

print("\n" + "="*60)
print("Таблица результатов")
print("="*60)
print(f"{'Датасет':<25} {'Метод':<15} {'N_clusters':<10} {'Silhouette':<12} {'Davies-Bouldin':<15} {'Calinski-Harabasz':<10}")
print("-"*90)

for r in results:
    print(f"{r['dataset']:<25} {r['method']:<15} {r['n_clusters']:<10} {r['silhouette']:<12.4f} {r['davies_bouldin']:<15.4f} {r['calinski_harabasz']:<10.2f}")

fig, axes = plt.subplots(3, 3, figsize=(15, 15))
row_names = ['D1 (исходный)', 'D2 (PCA)', 'D3 (t-SNE)']
col_names = ['K-Means', 'Agglomerative', 'DBSCAN']

for i, (dataset_name, X_data) in enumerate(datasets.items()):
    for j, (method_name, model) in enumerate(clustering_methods):
        labels = model.fit_predict(X_data)
        axes[i, j].scatter(X_data[:, 0], X_data[:, 1] if X_data.shape[1] > 1 else X_data[:, 0], 
                          c=labels, cmap='viridis', alpha=0.7, s=50)
        axes[i, j].set_title(f'{row_names[i]} - {col_names[j]}')
        axes[i, j].set_xlabel('признак 1')
        axes[i, j].set_ylabel('признак 2')

plt.tight_layout()
plt.show()

print("\nВыводы по кластеризации:")
print("="*60)

for dataset_name in datasets.keys():
    dataset_results = [r for r in results if r['dataset'] == dataset_name]
    best_method = max(dataset_results, key=lambda x: x['silhouette'] if x['silhouette'] > -1 else -2)
    print(f"\n{dataset_name}:")
    print(f"  Лучший метод: {best_method['method']} (silhouette = {best_method['silhouette']:.4f})")

d1_best = max([r for r in results if r['dataset'] == 'D1 (исходный, 4 признака)'], key=lambda x: x['silhouette'])
d2_best = max([r for r in results if r['dataset'] == 'D2 (PCA, 2 признака)'], key=lambda x: x['silhouette'])
d3_best = max([r for r in results if r['dataset'] == 'D3 (t-SNE, 2 признака)'], key=lambda x: x['silhouette'])

print(f"\nЛучший метод кластеризации для D1 (исходный): {d1_best['method']}")
print(f"Лучший метод кластеризации для D2 (PCA): {d2_best['method']}")
print(f"Лучший метод кластеризации для D3 (t-SNE): {d3_best['method']}")

print(f"\nЛучший метод снижения размерности для кластеризации: t-SNE")
print(f"  (на D3 silhouette = {d3_best['silhouette']:.4f}, что выше чем на D2: {d2_best['silhouette']:.4f})")
```
